# Re:Camp Colab + Drive + Blender setup

이 Notebook은 무료 Colab runtime에서 Blender 자동화 환경을 준비한다. Colab runtime은 임시이므로 `.blend`, FBX, 렌더, 리포트는 Google Drive에 저장한다. Blender GUI 수작업과 Unity Editor 작업은 이 Notebook의 범위가 아니다.

```python
from google.colab import drive
drive.mount('/content/drive')
print('Google Drive mounted.')
```

```python
from pathlib import Path
import shutil
import subprocess

REPO_URL = 'https://github.com/siri2677/re-camp.git'
TOOLS_REPO_URL = 'https://github.com/siri2677/re-camp-blender.git'
BRANCH = 'art/current-roster-gate-a-ch102'
REPO_DIR = Path('/content/re-camp')
TOOLS_DIR = Path('/content/re-camp-blender')
DRIVE_ROOT = Path('/content/drive/MyDrive/re-camp')
DRIVE_ROOT.mkdir(parents=True, exist_ok=True)
print(f'Drive root: {DRIVE_ROOT}')
print(f'Repository: {REPO_URL} / {BRANCH}')
print(f'Tools repository: {TOOLS_REPO_URL}')
```

```python
if not (REPO_DIR / '.git').exists():
    subprocess.run([
        'git', 'clone', '--depth', '1', '--branch', BRANCH, REPO_URL, str(REPO_DIR)
    ], check=True)
else:
    subprocess.run(['git', '-C', str(REPO_DIR), 'fetch', 'origin', BRANCH], check=True)
    subprocess.run(['git', '-C', str(REPO_DIR), 'checkout', BRANCH], check=True)
    subprocess.run(['git', '-C', str(REPO_DIR), 'pull', '--ff-only', 'origin', BRANCH], check=True)
if not (TOOLS_DIR / '.git').exists():
    subprocess.run([
        'git', 'clone', '--depth', '1', TOOLS_REPO_URL, str(TOOLS_DIR)
    ], check=True)
else:
    subprocess.run(['git', '-C', str(TOOLS_DIR), 'pull', '--ff-only'], check=True)
print('Repository ready:', REPO_DIR)
print('Tools ready:', TOOLS_DIR)
```

```python
import shutil

if shutil.which('blender') is None:
    subprocess.run(['apt-get', 'update', '-qq'], check=True)
    subprocess.run(['apt-get', 'install', '-y', 'blender'], check=True)

print(subprocess.run(['blender', '--version'], check=True, capture_output=True, text=True).stdout.splitlines()[0])
```

```python
for name in ('blend', 'exports', 'renders', 'reports', 'checkpoints'):
    (DRIVE_ROOT / name).mkdir(parents=True, exist_ok=True)

python_version = subprocess.run(['python', '--version'], capture_output=True, text=True)
print('Python:', (python_version.stdout or python_version.stderr).strip())
print('Git:', subprocess.run(['git', '--version'], capture_output=True, text=True).stdout.strip())
print('Drive workspace prepared:', DRIVE_ROOT)
```
